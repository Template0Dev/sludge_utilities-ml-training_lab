# ResNet Optuna tuning

This notebook defines paths and delegates tuning to the controller API.


```python
from pathlib import Path
import sys

from pyprojroot import here

PROJECT_ROOT = Path(here())
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from ml_training_lab.controller.resnet_tuning_controller import tune_resnet  # noqa: E402
from ml_training_lab.presentation.requests import ResNetRequest  # noqa: E402
```

```python
CONFIG_PATH = PROJECT_ROOT / "config/resnet_training.json"
DATASET_PATH = PROJECT_ROOT / "data/meta/interim/metadata_augmented.parquet"
IMAGE_ROOT = PROJECT_ROOT / "data/images"

request = ResNetRequest(
    project_root=PROJECT_ROOT,
    config_path=CONFIG_PATH,
    dataset_path=DATASET_PATH,
    image_root=IMAGE_ROOT,
)
request
```

```python
result = tune_resnet(request)

print(f"Tune id: {result.tune_id}")
print(f"Tune directory: {result.tune_dir}")
print(f"Optuna summary saved to: {result.summary_path}")
print(f"Optuna trials saved to: {result.trials_path}")
print(f"Optuna database: {result.study_database_path}")
```
