# CatBoost Optuna tuning

This notebook defines paths and delegates tuning to the controller API.


```python
from pathlib import Path
import sys

from pyprojroot import here

PROJECT_ROOT = Path(here())
sys.path.insert(0, str(PROJECT_ROOT / "src"))

from ml_training_lab.controller.catboost_tuning_controller import tune_catboost  # noqa: E402
from ml_training_lab.presentation.requests import CatBoostRequest  # noqa: E402
```

```python
CONFIG_PATH = PROJECT_ROOT / "config/gb_training.json"
BASE_DATASET_PATH = PROJECT_ROOT / "data/meta/interim/metadata_augmented.parquet"
EMBEDDING_DATASET_PATH = PROJECT_ROOT / "data/meta/processed/gb/metadata_dinov3_embeddings.parquet"

request = CatBoostRequest(
    project_root=PROJECT_ROOT,
    config_path=CONFIG_PATH,
    base_dataset_path=BASE_DATASET_PATH,
    embedding_dataset_path=EMBEDDING_DATASET_PATH,
)
request
```

```python
result = tune_catboost(request)
print(f"Tune id: {result.tune_id}")
print(f"Tune directory: {result.tune_dir}")
print(f"Optuna summary saved to: {result.summary_path}")
print(f"Optuna trials saved to: {result.trials_path}")
print(f"Optuna database: {result.study_database_path}")
```
